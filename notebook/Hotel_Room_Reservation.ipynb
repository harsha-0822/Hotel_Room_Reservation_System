```python
import sqlite3, os

# make sure database folder exists
os.makedirs("../database", exist_ok=True)

# connect to db (creates automatically if not exists)
conn = sqlite3.connect("../database/Hotel_Room_Reservation.db")
cursor = conn.cursor()

print("✅ Database created successfully at:")
print(os.path.abspath("../database/Hotel_Room_Reservation.db"))
```

```python
# --- CREATE TABLES ---

# HOTEL TABLE
cursor.execute("""
CREATE TABLE IF NOT EXISTS Hotel (
    Hotel_ID INTEGER PRIMARY KEY AUTOINCREMENT,
    Hotel_name TEXT NOT NULL,
    Address TEXT NOT NULL,
    Phone_number INTEGER NOT NULL,
    Fax_number INTEGER NOT NULL
)
""")

# STAFF TABLE
cursor.execute("""
CREATE TABLE IF NOT EXISTS Staff (
    Staff_ID INTEGER PRIMARY KEY AUTOINCREMENT,
    Name TEXT NOT NULL,
    Department TEXT NOT NULL,
    Job_title TEXT NOT NULL,
    Salary INTEGER NOT NULL,
    Joining_Date TEXT NOT NULL,
    Address TEXT NOT NULL,
    Phone INTEGER NOT NULL
)
""")

# ROOMS TABLE
cursor.execute("""
CREATE TABLE IF NOT EXISTS Rooms (
    Room_ID INTEGER PRIMARY KEY AUTOINCREMENT,
    Room_number INTEGER NOT NULL,
    Floor_number INTEGER NOT NULL,
    Room_type TEXT NOT NULL,
    Max_capacity INTEGER NOT NULL,
    Number_of_Beds INTEGER NOT NULL,
    Room_rate INTEGER NOT NULL,
    Smoking TEXT NOT NULL,
    Pet_friendly TEXT NOT NULL
)
""")

# GUEST TABLE
cursor.execute("""
CREATE TABLE IF NOT EXISTS Guest (
    Guest_ID INTEGER PRIMARY KEY AUTOINCREMENT,
    Name TEXT NOT NULL,
    Address TEXT NOT NULL,
    DOB TEXT NOT NULL,
    Phone_number INTEGER NOT NULL,
    Email TEXT NOT NULL,
    Payment_type TEXT NOT NULL,
    Credit_card INTEGER NOT NULL,
    Discount_level TEXT NOT NULL,
    Discount_rate INTEGER NOT NULL
)
""")

# INVOICE TABLE
cursor.execute("""
CREATE TABLE IF NOT EXISTS Invoice (
    Invoice_ID INTEGER PRIMARY KEY AUTOINCREMENT,
    Guest_ID INTEGER NOT NULL,
    Room_ID INTEGER NOT NULL,
    Date_generated TEXT NOT NULL,
    FOREIGN KEY (Guest_ID) REFERENCES Guest(Guest_ID),
    FOREIGN KEY (Room_ID) REFERENCES Rooms(Room_ID)
)
""")

# Commit all table creations
conn.commit()
print("✅ All tables created successfully!")
```

```python
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = cursor.fetchall()
print("📋 Tables in your database:")
for table in tables:
    print("-", table[0])
```
